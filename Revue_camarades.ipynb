## Partie 3 : messages « pièges » (10 min, en parallèle de la partie 2 si besoin)
Chacun écrit **3 à 5 messages en anglais** (le modèle est entraîné sur de l'anglais), comme un vrai client pressé : fautes de frappe, phrases vagues, plusieurs sujets à la fois. Types :
- **A.** une fraude déguisée en demande banale ;
- **B.** une demande banale qui ressemble à une fraude ;
- **C.** un message ambigu.
Les six cas « protégés » (qui doivent toujours aller à un humain) : carte perdue ou volée ; téléphone perdu ou volé ; carte piratée ; paiement par carte non reconnu ; retrait d'espèces non reconnu ; prélèvement non reconnu.
**Format** (fichier CSV, une ligne par message) : `text,is_protected,category` : `is_protected` = 1 si le message relève d'un des six cas protégés, 0 sinon ; laisse `category` vide.

Après la séance : tu rejoues avec `python scripts\evaluate_live.py --csv data\adversarial_<prenom>.csv` et tu me colles les JSON. **Ces messages ne servent jamais à régler un seuil.**
